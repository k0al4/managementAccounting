```python
# %%capture out
import matplotlib.pyplot as plt
import matplotlib
import numpy as np

# from ipywidgets import interactive, FloatSlider, IntSlider

# Initialize figure and axis
plt.ion() # causes plots to be plotted again when changed
# matplotlib.use('Qt5Agg')
# %matplotlib osx
fig, ax = plt.subplots()

# Define initial line data
x, y = [],[]

chart = ax.scatter(x,y)
plt.draw()

for i in range(10):
    x.append(np.random.rand(1)[0]*10)
    y.append(np.tan(x[-1]))
    chart.set_offsets(np.c_[x,y])
    fig.canvas.draw_idle()
    plt.pause(0.1)

# plt.waitforbuttonpress()
```
